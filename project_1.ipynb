```python


from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

splitter = RecursiveCharacterTextSplitter(chunk_size=400, chunk_overlap=50)
chunks = splitter.split_documents(pages)

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
db = FAISS.from_documents(chunks, embeddings)

print(f"✅ {len(chunks)} chunks ready - Barath resume RAG ready!")
```

```python
import os
from google import genai
from google.colab import userdata


try:

  api_key = userdata.get('GEMINI_API_KEY')
except:

  from getpass import getpass
  api_key = getpass("Enter your Gemini API Key:AQ. ")

client = genai.Client(api_key=api_key)
print("✅ Client ready securely!")

# Ippo mela sonna ask_resume function same
```

```python


def ask_resume(question):
    docs = db.similarity_search(question, k=3)
    context = "\n".join([d.page_content for d in docs])

    prompt = f"""
    You are a professional resume assistant for Barath Kumar.

    Context from resume:
    {context}

    Question: {question}

    Instructions:
    - Answer ONLY in clear, professional English.
    - Do not use Tamil or Tanglish.
    - Answer strictly based on the resume context.
    - If answer is not in resume, say "Not mentioned in the resume."
    - Be concise and professional.
    """

    response = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents=prompt
    )
    return response.text

print("Q1: What is my CGPA?")
print(ask_resume("What is my CGPA?"))
print("\n---\n")

print("Q2: What skills do I have for AI jobs?")
print(ask_resume("What skills do I have for AI jobs?"))
print("\n---\n")

print("Q3: What is my project?")
print(ask_resume("What is my project?"))
```

```python


def ask_resume_professional(question):
    docs = db.similarity_search(question, k=3)
    context = "\n".join([d.page_content for d in docs])

    prompt = f"""
    You are Barath Kumar's professional resume assistant.

    Resume Context:
    {context}

    User Question: {question}

    Guidelines:
    1. Respond only in professional, formal English.
    2. Provide concise, accurate, and resume-based answers.
    3. Do not use Tamil, Tanglish, or informal language.
    4. If the information is not in the resume, respond with "This information is not mentioned in the resume."
    5. Maintain a professional tone suitable for recruiters.
    """

    response = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents=prompt
    )
    return response.text

print("=== Barath Kumar - Resume AI Assistant ===")
print("Ask any question about the resume. Type 'exit' to quit.")
print("Examples: What is your educational background? | List your technical skills | Describe your project")
print("-" * 60)

while True:
    q = input("\nEnter your question: ")

    if q.lower().strip() == "exit":
        print("\nThank you. Session ended. Resume assistant closed.")
        break

    if not q.strip():
        continue

    print("\nAnswer:")
    print(ask_resume_professional(q))
    print("-" * 60)
```
